## Step 3: Handle Missing Values

**Identify:** Check how many missing values each column has.

**Strategy (applied generically, in case any column has nulls):**
- Numeric columns (`Sales`, `Quantity`, `Discount`, `Profit`, `Postal Code`) → fill with the **median**.
- Categorical columns (`Ship Mode`, `Region`, `State`, `City`) → fill with the **mode**.

This is the Kaggle "cleaned" version of the Superstore dataset, so it is expected to already
have zero missing values — but the code below is written to handle them generically so it
still works correctly on any messier version of the dataset.


```python
# Identify missing values
missing_before = df.isnull().sum()
print("Missing values per column:")
print(missing_before[missing_before > 0] if missing_before.sum() > 0 else "No missing values found in the dataset.")
```

```python
# Generic handling: fill numeric columns with median, categorical columns with mode
numeric_cols = df.select_dtypes(include=[np.number]).columns
categorical_cols = df.select_dtypes(include=['object']).columns

for col in numeric_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after cleaning:", df.isnull().sum().sum())
```

## Step 4: Basic Operations (Filter Rows, Select Columns)

Examples of common Pandas operations on the Superstore data.


```python
# Select specific columns
subset = df[['Order ID', 'Category', 'Sub-Category', 'Sales', 'Quantity', 'Profit']]
subset.head()
```

```python
# Filter rows: Technology orders with Sales > 500
tech_high_value = df[(df['Category'] == 'Technology') & (df['Sales'] > 500)]
print(f"Technology orders with Sales > 500: {len(tech_high_value)}")
tech_high_value.head()
```

```python
# Filter rows: loss-making orders (Profit < 0)
loss_orders = df[df['Profit'] < 0]
print(f"Orders that resulted in a loss: {len(loss_orders)}")
```

```python
# Filter rows: orders from the West region
west_orders = df[df['Region'] == 'West']
print(f"Orders from West region: {len(west_orders)}")
```

## Step 5: Remove Duplicates

Check for and remove any fully duplicated rows in the dataset.


```python
# Check duplicates
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows found: {duplicate_count}")
```

```python
# Remove duplicates (safe to run even if none exist)
df = df.drop_duplicates().reset_index(drop=True)
print(f"Shape after removing duplicates: {df.shape}")
```

## Step 6: Create a Derived Column

The Superstore dataset provides `Sales` as the total line-item revenue (i.e. price × quantity
already multiplied together) rather than a separate unit price column. To follow the
assignment's `total_amount = price * quantity` requirement in a meaningful way, we first
derive a `Unit_Price` column, then recompute `total_amount` from it — which also serves as a
validation check against the original `Sales` column.


```python
df['Unit_Price'] = round(df['Sales'] / df['Quantity'], 2)
df['total_amount'] = round(df['Unit_Price'] * df['Quantity'], 2)

df[['Sales', 'Quantity', 'Unit_Price', 'total_amount']].head()
```

```python
# Sanity check: total_amount should match the original Sales column (small rounding differences allowed)
diff = (df['total_amount'] - df['Sales']).abs()
print(f"Max difference between total_amount and Sales: {diff.max():.4f}")
```

## Step 7: Save the Cleaned Dataset

Export the cleaned DataFrame to a new CSV file.


```python
df.to_csv("superstore_cleaned.csv", index=False)
print("Cleaned dataset saved as 'superstore_cleaned.csv'")
print(f"Final shape: {df.shape}")
```
